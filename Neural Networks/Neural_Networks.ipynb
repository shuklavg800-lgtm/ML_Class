# Neural Networks

This notebook demonstrates a simple feedforward neural network for classification using `MLPClassifier` from scikit-learn.

Topics covered:

- Preparing a dataset
- Training a neural network
- Evaluating predictions
- Visualizing the decision boundary

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler

plt.style.use('seaborn-v0_8')
np.random.seed(42)
```

## 1. Create a Dataset

We will use a two-class nonlinear dataset generated with `make_moons()`.

```python
X, y = make_moons(n_samples=500, noise=0.20, random_state=42)

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=35)
plt.title('Input Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## 2. Split and Scale the Data

Neural networks usually perform better when features are scaled.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Training samples:', X_train.shape[0])
print('Testing samples:', X_test.shape[0])
```

## 3. Train the Neural Network

We will use one hidden layer with 10 neurons and the ReLU activation function.

```python
model = MLPClassifier(
    hidden_layer_sizes=(10,),
    activation='relu',
    solver='adam',
    max_iter=2000,
    random_state=42
)

model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
```

## 4. Evaluate the Model

```python
accuracy = accuracy_score(y_test, y_pred)
print('Accuracy:', round(accuracy, 3))
print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification Report:\n')
print(classification_report(y_test, y_pred))
```

## 5. Visualize the Decision Boundary

```python
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)

grid = np.c_[xx.ravel(), yy.ravel()]
grid_scaled = scaler.transform(grid)
Z = model.predict(grid_scaled).reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.35, cmap='coolwarm')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=35)
plt.title('Neural Network Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## 6. Training Loss Curve

The loss curve helps us see how the neural network improved during training.

```python
plt.figure(figsize=(8, 5))
plt.plot(model.loss_curve_)
plt.title('Training Loss Curve')
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.show()
```

## Summary

- A neural network learns nonlinear patterns from data.
- Feature scaling is important for stable training.
- Hidden layers help the model learn more complex decision boundaries.

You can experiment with:

- `hidden_layer_sizes=(20, 10)`
- different activations such as `'tanh'`
- larger or smaller noise values in the dataset
