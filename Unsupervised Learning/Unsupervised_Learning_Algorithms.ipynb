# Unsupervised Learning Algorithms

This notebook introduces a few common unsupervised learning techniques:

- K-Means Clustering
- Hierarchical Clustering
- DBSCAN
- Principal Component Analysis (PCA)

The examples use synthetic datasets so the notebook can run as-is.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.datasets import make_blobs, make_moons
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

plt.style.use('seaborn-v0_8')
np.random.seed(42)
```

## 1. K-Means Clustering

K-Means groups data points into `k` clusters by minimizing the distance between each point and its assigned cluster centroid.

```python
X_blobs, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.10, random_state=42)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_blobs)

print('K-Means silhouette score:', round(silhouette_score(X_blobs, kmeans_labels), 3))
print('Cluster centers:\n', kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=kmeans_labels, cmap='viridis', s=40)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', s=200, marker='X', label='Centroids')
plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

## 2. Hierarchical Clustering

Hierarchical clustering builds clusters step by step. In agglomerative clustering, every point starts as its own cluster and the algorithm merges the closest clusters iteratively.

```python
hierarchical = AgglomerativeClustering(n_clusters=4)
hier_labels = hierarchical.fit_predict(X_blobs)

print('Hierarchical clustering silhouette score:', round(silhouette_score(X_blobs, hier_labels), 3))
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=hier_labels, cmap='plasma', s=40)
plt.title('Agglomerative Hierarchical Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## 3. DBSCAN

DBSCAN is a density-based clustering algorithm. It can discover clusters of arbitrary shape and also mark noisy points as outliers.

```python
X_moons, _ = make_moons(n_samples=300, noise=0.08, random_state=42)
X_moons = StandardScaler().fit_transform(X_moons)

dbscan = DBSCAN(eps=0.25, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_moons)

num_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
num_noise = list(dbscan_labels).count(-1)

print('DBSCAN clusters found:', num_clusters)
print('Noise points:', num_noise)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_moons[:, 0], X_moons[:, 1], c=dbscan_labels, cmap='coolwarm', s=40)
plt.title('DBSCAN Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## 4. Principal Component Analysis (PCA)

PCA is a dimensionality reduction technique. It transforms the original features into a smaller set of new features called principal components while preserving as much variance as possible.

```python
X_high_dim, _ = make_blobs(n_samples=250, centers=4, n_features=5, cluster_std=2.0, random_state=42)
X_high_dim = StandardScaler().fit_transform(X_high_dim)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_high_dim)

print('Original shape:', X_high_dim.shape)
print('Reduced shape:', X_pca.shape)
print('Explained variance ratio:', pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], s=40, c='teal')
plt.title('PCA: Data Projected to 2 Dimensions')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

## Summary

- **K-Means** works well for roughly spherical clusters.
- **Hierarchical Clustering** is useful for understanding nested groupings.
- **DBSCAN** is good for irregular cluster shapes and outlier detection.
- **PCA** reduces dimensionality for visualization and preprocessing.

You can now experiment by changing hyperparameters such as `n_clusters`, `eps`, or `n_components`.
